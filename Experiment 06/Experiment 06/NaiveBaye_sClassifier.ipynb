```python
# Probability
yes = 9
no = 5
prob_yes = yes / (yes + no)
prob_no = no / (yes + no)
```

```python
new_Outlook = 'Sunny'
new_Temperature = 'Cool'
new_Humidity = 'High'
```

```python
yes_data = df[df['PlayTennis'] == 'Yes']
p_Outlook_yes = (yes_data['Outlook'] == new_Outlook).sum() / yes
p_Temperature_yes = (yes_data['Temperature'] == new_Temperature).sum() / yes
p_Humidity_yes = (yes_data['Humidity'] == new_Humidity).sum() / yes

yes_score = (
    prob_yes
    * p_Outlook_yes
    * p_Temperature_yes
    * p_Humidity_yes
)

no_data = df[df['PlayTennis'] == 'No']
p_Outlook_no = (no_data['Outlook'] == new_Outlook).sum() / no
p_Temperature_no = (no_data['Temperature'] == new_Temperature).sum() / no
p_Humidity_no = (no_data['Humidity'] == new_Humidity).sum() / no

no_score = (
    prob_no
    * p_Outlook_no
    * p_Temperature_no
    * p_Humidity_no
)
```

```python
if yes_score > no_score:
    print("Yes")
else:
    print("No")
```
