```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
path="/content/drive/MyDrive/ML Lab/Exp02/Salary.csv"
df=pd.read_csv(path)
```

```python
df.head(16)
```

PreProcessing

```python
#1.Filter(Not Required Does Not Have Irrelevent Features)
#2.Handling Non-Numeric Data(Not Required Because Data Set HAs Only Numeric Data)
#3.Handling Missing Data(#No Missing Data)

#4.Normalization Or Scaling(Required)
col=['SALARY']
df[col]=(df[col]-df[col].min())/(df[col].max()-df[col].min())
df.head(16)
```

**Simple** **Linear** **Regression**

Split Data

```python
X_train,X_test,Y_train,Y_test=train_test_split(
    df['YEARS'],df['SALARY'],test_size=0.2,random_state=42
)
```

Train Model

```python
model=LinearRegression()
model.fit(X_train.values.reshape(-1, 1),Y_train)
```

Prediction

```python
years = [[5]]
predicted_salary = model.predict(years)
print("Normalized Salary:", predicted_salary[0])

original_salary = predicted_salary*(df['SALARY'].max() - df['SALARY'].min()) + df['SALARY'].min()
print("Original Salary:",original_salary)
```

```python
y_pred = model.predict(X_test.values.reshape(-1, 1))

print("R² Score:", r2_score(Y_test, y_pred))
print("MSE:", mean_squared_error(Y_test, y_pred))
```
